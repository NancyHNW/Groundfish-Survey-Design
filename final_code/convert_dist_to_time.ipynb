```python
from data_loader import *
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# convert distance matrix to time values based on Dana's speeds
dist = load_true_dist()
time = np.zeros((np.shape(dist)[0], np.shape(dist)[1]))

for i in range(np.shape(dist)[0]):
    for j in range(np.shape(dist)[1]):
        # if i % 2 == 1 and i > 11 and j == i + 1:
        if i > 12 and j > 12 and ((i % 2 == 1 and j == i + 1) or (i % 2 == 0 and i == j + 1)):
            time[i, j] = dist[i, j] / 7
        else:
            time[i, j] = dist[i, j] / 14

np.save("local data/true_time.npy", time)
```
